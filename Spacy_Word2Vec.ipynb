```python
from gensim.models import Word2Vec, KeyedVectors
from tqdm.notebook import tqdm # Import tqdm for progress bar

text= "/content/cbow_s300.txt"
text_sg = "/content/skip_s300.txt"

# Use tqdm to provide a high-level progress bar for loading both models
with tqdm(total=2, desc="Loading Word2Vec Models") as pbar:
    pbar.set_description("Loading Skip-gram Model (skip_s300.txt)")
    modelo_sg = KeyedVectors.load_word2vec_format(text_sg)
    pbar.update(1) # Update after the first model loads

    pbar.set_description("Loading CBOW Model (cbow_s300.txt)")
    modelo = KeyedVectors.load_word2vec_format(text)
    pbar.update(1) # Update after the second model loads

print("Word2Vec models loaded successfully!")
```
